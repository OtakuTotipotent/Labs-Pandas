# GroupBy, Aggregation & Data Analysis

## Data

```python
import pandas as pd

data = {
    "Department": ["Sales", "IT", "Sales", "HR", "IT", "HR"],
    "Salary": [50000, 70000, 55000, 45000, 75000, 48000],
    "Experience": [2, 5, 3, 1, 6, 2],
}

df = pd.DataFrame(data)
print(df)
```

## Group By

```python
grouped_data = df.groupby("Department")
print(grouped_data)
```

### Calculate Average Salary

```python
average_salary = grouped_data["Salary"].mean()
print(average_salary)
```
