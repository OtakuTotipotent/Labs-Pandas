# GroupBy, Aggregation & Data Analysis

## Data

```python
import pandas as pd

data = {
    "Department": ["Sales", "IT", "Sales", "HR", "IT", "HR"],
    "Salary": [50000, 70000, 55000, 45000, 75000, 48000],
    "Experience": [2, 5, 3, 1, 6, 2],
}

df = pd.DataFrame(data)
print(df)
```

## Group By

```python
print(df.groupby("Department"))
```

### Calculate Average Salary

```python
print(df.groupby("Department")["Salary"].mean())
```

### Calculate Total Salary for Each Department

```python
print(df.groupby("Department")["Salary"].sum())
```

### Multiple Aggregation Functions

```python
print(df.groupby("Department")["Salary"].agg(["sum", "mean", "max", "min"]))
```
